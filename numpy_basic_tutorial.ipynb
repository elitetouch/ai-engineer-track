# 🔢 Basic Numpy Tutorial
This notebook introduces **NumPy**, the fundamental package for numerical computing in Python.
We'll cover array creation, indexing, reshaping, and basic mathematical operations.

```python
import numpy as np
```

## 1️⃣ Creating Arrays

```python
# From a Python list
arr = np.array([1, 2, 3, 4, 5])
arr
```

```python
# Multi-dimensional array
mat = np.array([[1, 2, 3], [4, 5, 6]])
mat
```

```python
# Zeros, Ones, and Random arrays
zeros = np.zeros((2,3))
ones = np.ones((3,2))
randoms = np.random.rand(2,2)
zeros, ones, randoms
```

## 2️⃣ Array Properties

```python
arr.shape, arr.dtype, arr.size, arr.ndim
```

## 3️⃣ Indexing and Slicing

```python
# Indexing
arr[0], arr[-1]

# Slicing
arr[1:4]
```

## 4️⃣ Reshaping and Combining

```python
a = np.arange(1, 10)
reshaped = a.reshape((3, 3))

b = np.array([[10, 20, 30]])

# Stack arrays vertically and horizontally
v_stack = np.vstack([reshaped, b])
h_stack = np.hstack([reshaped, reshaped])
v_stack, h_stack
```

## 5️⃣ Basic Mathematical Operations

```python
# Element-wise operations
arr + 10
arr * 2

# Aggregate functions
arr.sum(), arr.mean(), arr.min(), arr.max(), arr.std()
```

## 6️⃣ Boolean Masking and Filtering

```python
# Filter elements greater than 3
mask = arr > 3
filtered = arr[mask]
filtered
```
