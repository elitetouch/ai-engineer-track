# 🐼 Basic Data Wrangling with Pandas
This notebook covers basic data wrangling operations using **pandas** — loading data, exploring, cleaning, filtering, grouping, and summarizing datasets.

```python
# Import essential libraries
import pandas as pd
import numpy as np
```

## 1️⃣ Load Data
You can load data from CSV, Excel, or JSON. Let's use a CSV example.

```python
# Load a sample CSV file
# Replace 'data.csv' with your dataset path
df = pd.read_csv('data.csv')

# Display the first 5 rows
df.head()
```

## 2️⃣ Inspect Data

```python
# Check basic info
df.info()

# Summary statistics
df.describe()

# Check column names
df.columns
```

## 3️⃣ Handle Missing Values

```python
# Check missing values
df.isnull().sum()

# Fill missing values
df['column_name'] = df['column_name'].fillna('Unknown')

# Drop rows with missing values
df = df.dropna()
```

## 4️⃣ Filter and Select Data

```python
# Select columns
df[['column1', 'column2']].head()

# Filter rows based on a condition
filtered = df[df['column_name'] > 50]

# Sort values
df.sort_values('column_name', ascending=False).head()
```

## 5️⃣ Create or Modify Columns

```python
# Create a new column
df['total'] = df['quantity'] * df['price']

# Rename columns
df.rename(columns={'old_name': 'new_name'}, inplace=True)

# Drop a column
df.drop('unwanted_column', axis=1, inplace=True)
```

## 6️⃣ Grouping and Aggregation

```python
# Group by one column and calculate mean
grouped = df.groupby('category')['total'].mean()

# Multiple aggregations
agg_df = df.groupby('category').agg({'price': 'mean', 'quantity': 'sum'})
agg_df.head()
```

## 7️⃣ Save Cleaned Data

```python
# Save cleaned data to new CSV
df.to_csv('cleaned_data.csv', index=False)
```

✅ Done! You now know the basics of data wrangling using pandas.
